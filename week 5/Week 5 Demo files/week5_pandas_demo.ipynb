## Step 1 — Load the data and take a first look

`pd.read_csv()` loads a CSV file into a DataFrame. `df.head()` shows the first five rows. This is always your first move.

### What is a DataFrame?

A **DataFrame** is pandas' main way of storing tabular data. You can think of it like a spreadsheet or a CSV loaded into Python: it has rows, columns, and column names.

Pandas uses DataFrames because they make it easier to inspect, filter, summarize, clean, and analyze data with short commands. Instead of manually looping through every row in a CSV file, you can ask questions like:

```python
df.head()
df['rating'].mean()
df[df['rating'] < 3]
```

In this notebook, `df` is the name of our DataFrame. Once the CSV is loaded into `df`, every later step uses that DataFrame to explore the dataset.

````python
df = pd.read_csv('app_reviews_demo.csv')
df.head()
````

`df.tail()` shows the last five rows — useful for spotting truncated exports or checking whether the file ended cleanly.

````python
df.tail()
````

## Step 4 — Find the missing values — `df.isnull().sum()`

`df.isnull()` returns a DataFrame of True/False — True wherever a value is missing. `.sum()` counts the Trues per column.

In plain English you might ask the agent: 
>“For each column in the DataFrame, count how many cells are empty or missing.”

**Why this matters before any analysis:** if you want to group things, you might run a `groupby()` on a column that has nulls. Pandas silently drops those rows. You may not notice — but your results will be wrong.

````python
df.isnull().sum()
````

### Track rows with missing `device_type`

Every row where `device_type` is null is copied into a separate table called `null_device_type`. That leaves the main `df` unchanged and gives you an explicit audit list of those reviews (IDs, apps, ratings, etc.) for follow-up or export.

````python
null_device_type = df.loc[df["device_type"].isna()].copy().reset_index(drop=True)
print("rows with missing device_type:", len(null_device_type))
null_device_type
````

### New table — rows that are **not** missing `device_type`

The main `df` still includes every row. This table keeps **only** reviews where `device_type` has a value (rows with NaN in that column are dropped). Use it when you want to analyze or export just reviews with a known device, separate from `null_device_type` and from the full `df`.

The variable below is called `df_not_missing_device_type`. Equivalent filter: `df[df["device_type"].notna()]`.

````python
df_not_missing_device_type = df.dropna(subset=["device_type"]).copy().reset_index(drop=True)
print("full df rows:", len(df), "| not missing device_type:", len(df_not_missing_device_type))
df_not_missing_device_type.head()
````

### Statistical test: is **rating** different when `device_type` is missing?

Comparing means is a start. To ask whether a gap is plausibly due to **chance**, we run hypothesis tests. **Ratings are ordinal (1–5)**, so treat the Mann–Whitney result as the primary read; Welch’s *t*-test assumes ratings behave like continuous numbers (common but stronger). The **chi-square** test asks whether the **whole rating distribution** (1s, 2s, …) differs between rows with vs without a device label—not only the average.

**How to read *p*-values:** if *p* is small (often below 0.05), you would rarely see a gap this large if the two groups were truly identical—that supports a **statistically significant** difference. It does **not** prove why the gap exists (confounding is possible). Always report **sample sizes** and **descriptive stats** next to *p*.

The next cell uses **only** `pandas`, `numpy`, and the standard-library `statistics` module (no SciPy), so it should run in any environment where your Week 5 notebook already runs.

````python
import numpy as np
from scipy import stats

rating_missing = df.loc[df["device_type"].isna(), "rating"]
rating_known = df.loc[df["device_type"].notna(), "rating"]

n_m, n_k = len(rating_missing), len(rating_known)
print("Sample sizes")
print("  missing device_type:", n_m)
print("  known device_type:  ", n_k)
print()
print("Rating — descriptive")
print("  missing:  mean", round(rating_missing.mean(), 3), "  median", rating_missing.median(), "  std", round(rating_missing.std(), 3))
print("  known:    mean", round(rating_known.mean(), 3), "  median", rating_known.median(), "  std", round(rating_known.std(), 3))
print("  difference in means (missing − known):", round(rating_missing.mean() - rating_known.mean(), 3))
print()

# Ordinal-friendly: same distribution of ranks? (two-sided)
u_stat, p_mannwhitney = stats.mannwhitneyu(
    rating_missing, rating_known, alternative="two-sided"
)
print("Mann–Whitney U (non-parametric, good for 1–5 ratings)")
print("  U =", round(u_stat, 2), "   p-value =", round(p_mannwhitney, 5))
print()

# Parametric comparison (treats stars as numeric)
t_stat, p_welch = stats.ttest_ind(rating_missing, rating_known, equal_var=False)
print("Welch’s t-test (numeric shortcut; use with caution on Likert data)")
print("  t =", round(t_stat, 3), "   p-value =", round(p_welch, 5))
print()

pooled_sd = np.sqrt(
    ((n_m - 1) * rating_missing.var(ddof=1) + (n_k - 1) * rating_known.var(ddof=1))
    / (n_m + n_k - 2)
)
cohen_d = (rating_missing.mean() - rating_known.mean()) / pooled_sd
print("Effect size — Cohen’s d (missing − known, pooled SD):", round(cohen_d, 3))
print("  (rough guides: small ~0.2, medium ~0.5, large ~0.8)")
print()

# Full distribution: missing vs not × rating categories
ct = pd.crosstab(df["device_type"].isna(), df["rating"], rownames=["missing_device_type"])
chi2, p_chi, dof, _ = stats.chi2_contingency(ct)
print("Chi-square test (does the rating *distribution* differ?)")
print("  χ² =", round(chi2, 3), "   df =", dof, "   p-value =", round(p_chi, 5))
print()
print("Counts: rows = missing device_type?, columns = rating")
print(ct)
print()

alpha = 0.05
print(f"Conclusion at α = {alpha} (typical teaching threshold)")
for label, p in [
    ("Mann–Whitney", p_mannwhitney),
    ("Welch t", p_welch),
    ("Chi-square", p_chi),
]:
    verdict = "reject H₀ — difference is statistically significant" if p < alpha else "do not reject H₀ — not significant at this α"
    print(f"  {label}: p = {p:.5f} → {verdict}")
````

### So what? 
Good question. Finding missing values is only the first step. The “so what?” is: missing data can change your results, especially when you group or summarize.

If `device_type` has 63 missing values (which we now know it does!), you have a few reasonable choices:

1. **Keep them as their own category**
   
     Plain English for the agent: 
     >“Treat missing device types as `Unknown` so they still show up in the analysis.” This is often the best first move because it makes the missingness visible.

```python
     df['device_type'] = df['device_type'].fillna('Unknown')
     df.groupby('device_type')['rating'].mean()
```

2. **Drop rows with missing values**
   
     Plain English for the agent: 
     >“Only analyze rows where `device_type` is known.” This is okay if the missing rows are small in number and not important to the question. But here, 63 out of 500 is about 13%, so dropping them could affect the result.

```python
     df_clean = df.dropna(subset=['device_type'])
     df_clean.groupby('device_type')['rating'].mean()
```

3. **Ask whether the missingness matters**
   
     Before deciding, compare the missing rows to the non-missing rows. Plain English:
     > “Do reviews with missing device type have different ratings than reviews with known device type?”

```python
     df[df['device_type'].isnull()]['rating'].mean()
     df[df['device_type'].notnull()]['rating'].mean()
```
If they look different, dropping them may bias your analysis.

### A good rule of thumb:

_If the missing column is important to your question, don’t silently drop it. Either keep missing values as `Unknown`, or explain why you removed them._

## Step 5 — Distribution of a column — `value_counts()`

`value_counts()` counts how many times each unique value appears. It's the most useful single operation for categorical data.

Use it to answer: *What are the most common values? Is the data balanced or skewed toward one category?*

````python
# How many reviews per app?
df['app'].value_counts()
````

````python
df['device_type'].value_counts(dropna=False)
````

### What's the distribution of ratings?

````python
df['rating'].value_counts().sort_index()
````

---
## Step 6 — Filter to a subset — boolean indexing

`df[condition]` returns only the rows where the condition is True. The condition is written as a comparison inside the brackets.

This is the same as Excel's filter — but repeatable, stackable, and documented in your code.

Ask the agent:
>“Using pandas, show me only the reviews where rating is less than 2. Use df[condition] syntax and explain what the filtered DataFrame contains.”

````python
# Reviews with a rating of 2 or below
low_rated = df[df['rating'] <= 2]

print(f"{len(low_rated):,} low-rated reviews ({len(low_rated)/len(df):.1%} of total)")
low_rated.head()
````

You can stack two conditions with & (and) or | (or). Parentheses are required around each condition.

````python
df[(df['rating'] <= 2) & (df['app'] == 'Fieldkit')].head(10)
````

---
## Step 7 — Compare across groups — `groupby()`

`groupby()` splits the DataFrame by a categorical column, then applies an aggregation function (mean, count, sum, etc.) to a numeric column within each group.

This answers the sprint question: *How does [metric] vary by [category]?*

````python
# Average rating per app — sorted lowest to highest
df.groupby('app')['rating'].mean().round(2).sort_values()
````

````python
# More than one stat at once with .agg()
df.groupby('app')['rating'].agg(['mean', 'count', 'std']).round(2)
````

````python
# Groupby across two columns
df.groupby(['app', 'device_type'])['rating'].mean().round(2)
````

## CourtListener bridge — the five in-class operations on legal-search data

This section uses the **same five operations** from class (`head` / `info`, `value_counts`, boolean filter, `groupby` + `mean`, `isnull`) on a small **CourtListener-shaped** CSV (`courtlistener_firstlook_sample.csv`). It mirrors what you build in `week5_courtlistener_cases.ipynb` when you stack API pulls into `df_mp1` — here you can practice offline without a token.


````python
from IPython.display import display

# --- Five class operations on CourtListener-shaped rows (see week5_courtlistener_cases.ipynb for live API) ---
cl = pd.read_csv('courtlistener_firstlook_sample.csv')

# 1) df.head() and df.info() — what does the data look like?
display(cl.head())
cl.info()

# 2) df['column'].value_counts() — what are the most common values?
print("\nRows per dataset slice (miranda vs seattle federal vs king county appellate):")
print(cl['dataset'].value_counts())

# 3) df[df['column'] > value] — filter to a subset (decisions on or after 1990-01-01)
cl['_decision_dt'] = pd.to_datetime(cl['date_of_decision'], errors='coerce')
recent = cl[cl['_decision_dt'] >= pd.Timestamp('1990-01-01')]
print(f"\nFiltered to decisions on or after 1990-01-01: {len(recent)} rows")
display(recent.head())

# 4) df.groupby('column')['other'].mean() — group and summarize mean cite_count by dataset slice
print("\nMean cite_count by dataset:")
print(cl.groupby('dataset')['cite_count'].mean().round(2))

# 5) df.isnull().sum() — find missing data (judge is often sparse in API exports)
print("\nMissing values per column:")
print(cl.isnull().sum())
````

## The first-look checklist — summary

```python
df = pd.read_csv('your_file.csv')   # load

df.head()                            # first five rows
df.tail()                            # last five rows
df.info()                            # columns, types, non-null counts
df.describe()                        # numeric summary stats
df.describe(include='all')           # includes categoricals

df.isnull().sum()                    # missing value count per column
df.isnull().mean().mul(100).round(1) # missing as percentage

df['col'].value_counts()             # distribution of a categorical column
df['col'].value_counts(dropna=False) # include NaN in the count

df[df['col'] > value]                # filter rows
df[(df['col1'] > v1) & (df['col2'] == v2)]  # stacked filter

df.groupby('cat_col')['num_col'].mean()      # group and aggregate
df.groupby('cat_col')['num_col'].agg(['mean','count','std'])
```

Run these in order. Every time. On every new dataset.
